```python
import numpy as np 
import cv2
import dlib
from imutils import face_utils
```

```python
#shape rpedictor 
shape_predictor= dlib.shape_predictor('./models/shape_predictor_68_face_landmarks.dat')

#shape_description
shape_description = dlib.face_recognition_model_v1('./models/dlib_face_recognition_resnet_model_v1.dat')
```

```python
img = cv2.imread('./USman_model_data/usman_img_8.jpg')
img = cv2.resize(img,None,fx=0.2,fy=0.2,interpolation=cv2.INTER_AREA)
```

```python
image=img.copy()
face_detector= dlib.get_frontal_face_detector()

faces = face_detector(image)
print(faces)
```

```python
for box in faces:
    pt1= box.left(), box.top()
    pt2= box.right(), box.bottom()

    face_shape= shape_predictor(image,box)
    face_shape_array= face_utils.shape_to_np(face_shape)
    face_discriptor= shape_description.compute_face_descriptor(image,face_shape)
    for points in face_shape_array:
        cv2.circle(image,tuple(points),3,(0,255,255))
    cv2.rectangle(image,pt1,pt2,(0,255,0))


cv2.imshow('face', image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
