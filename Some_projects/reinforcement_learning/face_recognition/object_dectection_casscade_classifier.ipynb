```python
face_cade= cv2.CascadeClassifier('./casscade_clasifier/haarcascade_frontalface_default.xml')
eye_cade= cv2.CascadeClassifier('./casscade_clasifier/haarcascade_eye.xml')

smile_cascade= cv2.CascadeClassifier('./casscade_clasifier/haarcascade_smile.xml')
```

```python
img2= cv2.imread('./USman_model_data/get_from_internet.jpg')
img2= cv2.resize(img2,None,fx=0.5,fy=0.5)
    
# cv2.imshow('friends',img1)
# cv2.waitKey(0)
# cv2.destroyAllWindows()
```

```python
faces,num_detections=face_cade.detectMultiScale2(img2,minNeighbors=10)
```

```python
for x,y,w,h in faces:
    face_roi= img2[y:y+h,x:x+w].copy()
    cv2.rectangle(img2,(x,y),(x+w,y+h),(0,255,0),2)

    eyes, num_eyes= eye_cade.detectMultiScale2(face_roi,minNeighbors=4)
    for ex,ey,ew,eh in eyes:
        cx= x+ex+ew//2
        cy= y+ey+eh//2
        r= eh//2
        cv2.circle(img2,(cx,cy),r,(0,0,255),2)
    smile, num_smiles= smile_cascade.detectMultiScale2(face_roi,minNeighbors=10)
    for sx,sy,sw,sh in smile:
        cv2.rectangle(img2,(x+sx,y+sy),(x+sx+sw,sy+y+sh),(255,255,0),2)




cv2.imshow('friends',img2)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
