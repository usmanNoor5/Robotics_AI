```python
import numpy as np 
import cv2
```

```python
img= cv2.imread('./USman_model_data/get_from_internet.jpg')
img= cv2.resize(img,None,fx=0.5,fy=0.5)
```

```python
img.shape
```

```python
net = cv2.dnn.readNetFromCaffe('./models/deploy.prototxt.txt','./models/res10_300x300_ssd_iter_140000.caffemodel')
```

```python
blob = cv2.dnn.blobFromImage(img,1,(300,300),(104,177,123),swapRB=False)
```

```python
net.setInput(blob)
detection=net.forward()
```

```python
detection.shape
```

```python
h,w= img.shape[:2]
for i in range(0,detection.shape[2]):
    confindence= detection[0,0,i,2]
    if confindence>=0.5:
        box= detection[0,0,i,3:7]
        box= box*np.array([w,h,w,h]) 
        box= box.astype(int)
        sx,sy,ex,ey=box
        cv2.rectangle(img,(sx,sy),(ex,ey),(0,255,0),2)
        text = 'Face:{:.2f} %'.format(confindence*100)
        cv2.putText(img,text,(sx,sy-10),cv2.FONT_HERSHEY_PLAIN,1,(0,0,0))
    
cv2.imshow('face_detected', img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
