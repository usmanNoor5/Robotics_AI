```python
import numpy as np

fs = 10.1725  # sampling frequency (Hz)

n_init_rest  = int(30 * fs)   # 30 sec
n_grip       = int(10 * fs)   # 10 sec
n_rest       = int(20 * fs)   # 20 sec
n_final_rest = int(30 * fs)   # 30 sec
```

```python
def split_channel_trials(df, channel_name):
    """
    df: one subject DataFrame (shape ~ [time, 20])
    channel_name: e.g. 'ch1'
    
    Returns: dictionary with named segments
    """
    
    segments = {}
    idx = 0
    
    # ✅ Initial Rest
    segments[f"initial_rest_{channel_name}"] = df[channel_name].iloc[idx:idx + n_init_rest].values
    idx += n_init_rest
    
    # ✅ 10 Trials (Grip + Rest)
    for i in range(1, 11):
        # Grip
        segments[f"grip_{i}_{channel_name}"] = df[channel_name].iloc[idx:idx + n_grip].values
        idx += n_grip
        
        # Rest
        segments[f"rest_{i}_{channel_name}"] = df[channel_name].iloc[idx:idx + n_rest].values
        idx += n_rest
    
    # ✅ Final Rest
    segments[f"final_rest_{channel_name}"] = df[channel_name].iloc[idx:idx + n_final_rest].values
    
    return segments
```

```python
def split_all_channels_one_subject_pd(df):
    """
    df: one subject DataFrame (time x channels)
    Returns: dictionary of Pandas Series
    """
    
    all_segments = {}
    
    for col in df.columns:   # ch1 → ch20
        idx = 0
        
        # ✅ Initial Rest (Pandas Series)
        all_segments[f"initial_rest_{col}"] = df[col].iloc[idx:idx + n_init_rest]
        idx += n_init_rest
        
        # ✅ 10 Trials
        for i in range(1, 11):
            all_segments[f"grip_{i}_{col}"] = df[col].iloc[idx:idx + n_grip]
            idx += n_grip
            
            all_segments[f"rest_{i}_{col}"] = df[col].iloc[idx:idx + n_rest]
            idx += n_rest
        
        # ✅ Final Rest
        all_segments[f"final_rest_{col}"] = df[col].iloc[idx:idx + n_final_rest]
    
    return all_segments
```
