```python
import torch
import torchvision
import torchvision.transforms as transform
import numpy as np 
```

```python
tensor_transform= transform.Compose([transform.ToTensor()])
train_data= torchvision.datasets.CIFAR10(root= './data', download=True,train=True,transform=tensor_transform)

val_data= torchvision.datasets.CIFAR10(root= './data', download=True,train=False,transform=tensor_transform)
```

```python
train_data, val_data
```

```python
cifar10_labels = {
    0: "airplane",
    1: "automobile",
    2: "bird",
    3: "cat",
    4: "deer",
    5: "dog",
    6: "frog",
    7: "horse",
    8: "ship",
    9: "truck"
}
```

```python
train_data_lodder= torch.utils.data.DataLoader(train_data,shuffle=True,batch_size=4)
val_data_lodder= torch.utils.data.DataLoader(val_data,shuffle=True,batch_size=4)
```

```python
train_iter= iter(train_data_lodder)
images,lables=next(train_iter)
# images, labels = train_iter.next()
```

```python
print("Images shape:", images.shape)   # torch.Size([4, 3, 32, 32])
print("Labels:", lables)              # tensor of 4 labels
```

```python
import matplotlib.pyplot as plt 
def imgshow(img):
    img_np= img.numpy()
    plt.imshow(np.transpose(img_np,(1,2,0)))
    plt.show()

plt.figure(figsize=(4, 4))

imgshow(torchvision.utils.make_grid(images))
```
